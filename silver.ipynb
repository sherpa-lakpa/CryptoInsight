```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
from pyspark.sql.functions import col

df = spark.read.format("delta").load("/mnt/bronze/crypto")
# df = spark.read.format("csv").option("header", "true").load("bronze")
df_clean = df.select(
    col("coin"),
    col("date"),
    col("price")
)

df_clean = df_clean.dropna()
df_clean.show(5)
```

```python
df_clean.write.format("delta") \
    .mode("overwrite") \
    .save("/mnt/silver/crypto")
# df_clean.write.option("header", "true").format("csv").save("silver")
```
