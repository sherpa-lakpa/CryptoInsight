```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
from pyspark.sql.functions import avg

# df = spark.read.format("delta").load("/mnt/silver/crypto")
# df = spark.read.format("csv").option("header", "true").load("silver")
df = spark.table("workspace.cryptoinsight.silver")

df_gold = df.groupBy("coin") \
    .agg(avg("price").alias("avg_price"))
df_gold.show(5)
```

```python
# df_gold.write.format("delta") \
#     .mode("overwrite") \
#     .save("/mnt/gold/crypto")
# df_gold.write.option("header", "true").format("csv").save("gold")
```

```python
df_gold.write.partitionBy("coin") \
    .format("delta") \
    .mode("append") \
    .saveAsTable("workspace.cryptoinsight.gold")
```
