```python
from utils.spark_utils import get_spark
from utils.logger import get_logger
from pyspark.sql.functions import avg, col, to_date, month

spark = get_spark()
logger = get_logger()
logger.info("Starting Gold Monthly Layer")
```

```python
import json

with open("../../configs/entities.json") as f:
    config = json.load(f)

config = config['entities'][0]
```

```python

df = spark.read.table(config["silver_table"])

df_gold = df.withColumn("month", month("ingestion_time")) \
        .groupBy("coin", "month") \
        .agg(avg("price").alias("avg_price"))
```

```python
df_gold.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable(config["gold_tables"]["monthly"])
```

```python
logger.info("Gold Monthly Load Complete")
```
